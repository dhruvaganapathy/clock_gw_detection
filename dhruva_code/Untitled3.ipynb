```python
%matplotlib notebook
# or %matplotlib widget for ipympl (requires ipympl installed)
import matplotlib.pyplot as plt
import numpy as np

from matplotlib.animation import FuncAnimation

# --- Constants ---
AU = 1.496e11       # meters
day = 24*3600       # seconds
n_days = 7          # number of days to simulate
dt = 0.01           # fraction of a day per frame
n_frames = int(n_days/dt)

# Mean motion at 1 AU (radians per day)
n = 2 * np.pi / 365.25  

# --- Satellite setup ---
phase_offsets_deg = [0, 3.83, 180, 183.83, 90, 93.83]  # satellites along orbit
phase_offsets = np.deg2rad(phase_offsets_deg)
radii = np.ones(6) * AU

def satellite_positions(t):
    theta = n * t + phase_offsets
    x = radii * np.cos(theta)
    y = radii * np.sin(theta)
    return x, y

# --- Plot setup ---
fig, ax = plt.subplots(figsize=(6,6))
ax.set_aspect('equal')
ax.set_xlim(-1.2*AU, 1.2*AU)
ax.set_ylim(-1.2*AU, 1.2*AU)
ax.plot(0,0,'yo',markersize=12,label='Sun')

sat_dots, = ax.plot([],[], 'bo', markersize=6)
baseline_lines = [ax.plot([],[], 'r-')[0] for _ in range(3)]
ax.set_title("6 satellites, 3 baselines at 1 AU")

# --- Animation function ---
def update(frame):
    t = frame * dt  # in days
    x, y = satellite_positions(t)
    sat_dots.set_data(x, y)
    pairs = [(0,1),(2,3),(4,5)]
    for i, (a,b) in enumerate(pairs):
        baseline_lines[i].set_data([x[a], x[b]], [y[a], y[b]])
    return [sat_dots, *baseline_lines]

from IPython.display import HTML

ani = FuncAnimation(fig, update, frames=n_frames, interval=50, blit=True)
HTML(ani.to_jshtml())
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Constants ---
AU = 1.496e11       # meters
day = 24*3600       # seconds
n_days = 7          # number of days to simulate
dt = 0.1            # days per timestep
t = np.arange(0, n_days+dt, dt)  # time array in days

# Mean motion at 1 AU (radians per day)
n = 2 * np.pi / 365.25  

# --- Satellite setup ---
# 6 satellites, 3 baselines (disjoint)
# Baseline length ~10^10 m, phase separation ~4 deg
phase_offsets_deg = [0, 3.83, 180, 183.83, 90, 93.83]  # satellites along orbit
phase_offsets = np.deg2rad(phase_offsets_deg)
radii = np.ones(6) * AU

# --- Satellite positions as function of time ---
def satellite_positions(ti):
    theta = n*ti + phase_offsets
    x = radii * np.cos(theta)
    y = radii * np.sin(theta)
    return x, y

# --- Compute baseline angles ---
pairs = [(0,1),(2,3),(4,5)]
angles = np.zeros((len(t), len(pairs)))  # store angles in degrees

for i, ti in enumerate(t):
    x, y = satellite_positions(ti)
    for j, (a,b) in enumerate(pairs):
        dx = x[b] - x[a]
        dy = y[b] - y[a]
        angles[i,j] = np.degrees(np.arctan2(dy, dx))

# --- Plot angles ---
plt.figure(figsize=(8,5))
for j in range(len(pairs)):
    plt.plot(t, angles[:,j], label=f'Baseline {j+1} (S{pairs[j][0]+1}-S{pairs[j][1]+1})')

plt.xlabel("Time [days]")
plt.ylabel("Baseline angle w.r.t inertial frame [deg]")
plt.title("Baseline angles relative to distant stars over a week")
plt.legend()
plt.grid(True)

```
